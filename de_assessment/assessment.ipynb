```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_json('fake_property_data.json')
```

```python
print(df['Reviewed_Status'].value_counts())
print(df['Reviewed_Status'].dtype)
```

```python
df['Reviewed_Status'] = df['Reviewed_Status'].str.strip()
df['Reviewed_Status'].replace('Neww', 'New', inplace=True)
df['Reviewed_Status'].replace('', np.nan, inplace=True)
```

```python
df['Most_Recent_Status'] = df['Most_Recent_Status'].str.strip()
df['Most_Recent_Status'].replace('Close', 'Closed', inplace=True)
df['Most_Recent_Status'].replace('pending', 'Pending', inplace=True)
df['Most_Recent_Status'].replace('Always Kick', 'Kick', inplace=True)
df['Most_Recent_Status'].replace('', np.nan, inplace=True)
```

```python
print(df['Source'].value_counts())
print(df['Source'].dtype)
```

```python
df['Source'] = df['Source'].str.strip()
df['Source'].replace(['M L S', 'M.L.S.'], 'MLS', inplace=True)
```

```python
records_val = []
for id, val_list in df[['property_id', 'Valuation']].itertuples(index=False):         
    if isinstance(val_list, list):
        for d in val_list:   
            records_val.append({'property_id': id, **d})            
    

df_valuation = pd.DataFrame(records_val)

# df.itertuples(index=False) returns each row as a tuple and if index = True then index bi return ho jaega tuple me
# **d for unpacking key-value pair
```
