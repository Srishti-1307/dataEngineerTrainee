## Difference between Python Lists and Numpy Arrays

```python
import numpy as np
```

### Memory used

```python
list1= list(range(100))
```

```python
arr1= np.arange(100)
```

```python
import sys
```

```python
sys.getsizeof(1)*len(list1)              #getsizeof(koi sa bi element list me se (qki sbka size same hai) * no. of elements in list (which is 100))
```

```python
arr1.itemsize*arr1.size                 #arr1.itemsize se arr1 ke each element ka size (1 baar) * no. of elements in arr1 (which is 100)
```

### Time Consumed

```python
import time
```

```python
x= list(range(100000))
y= list(range(100000, 200000))

start_time_l= time.time()

c= [x+y for x,y in zip(x,y)]

print(time.time()-start_time_l)
```

```python
a= np.arange(100000)
b= np.arange(100000,200000)

start_time_a= time.time()

c= a+b

print(time.time()- start_time_a)
```
