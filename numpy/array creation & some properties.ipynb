## Numpy/nd-Array Creation Methods

1) np.array()

2) np.zeros/ones/empty/random()

3) np.arange() 

4) np.linspace()

5) copy()

6. identity()

```python
import numpy as np
```

```python
arr1= np.array([1,3,0,7,2,0,0,3])       
arr1
```

```python
arr2= np.array([[2,1,0,1],[2,9,1,1],[1,6,0,8]])
arr2
```

```python
arr3= np.zeros((2,4), dtype='int32')
arr3
```

```python
arr4= np.ones((5,4))
arr4
```

```python
arr5= np.identity(6)
arr5
```

```python
arr6= np.arange(2,14,2)                                    #np.arange(kaha se, kaha tk, step-size)
arr6
```

```python
arr7= np.linspace(1,10,20)                                  #np.linspace(kaha se, kaha tk, kitne)
arr7
```

```python
arr8= arr7.copy()                                        #A copy of arr7 array
arr8
```

```python
arr10= np.random.randint(low=1, high=25, size=20).reshape(4,5)             
arr10
```

```python
# you can use "np.random.randint(1,25,20)" this instead
# np.random.rand(10)                -- Generates random numbers from a uniform distribution between 0 and 1.
# np.random.randint(1,10,15)       -- Generates random integers within a specified range.
#a= np.random.randn(10)           -- Generates 10 random numbers from normal distribution
```
