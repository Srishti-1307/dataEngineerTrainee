```python
import numpy as np
```

```python
a1= np.arange(9).reshape(3,3)
a2= np.array([0,1,2])
```

```python
print(a1, a2)
```

```python
a1+a2
```

#### Above you saw that.... a1 is 3x3 and a2 is 1x3 still addition is being possible. This is due to Broadcasting

### Broadcasting......??   The smaller array is upgraded to the size of larger array so that they have compatible shapes. 

##### if broadcasting was't there... we we would be restricted to perform operations on same-sized arrays. 
Below we'll see where the Broadcasting is allowed...

```python
arr1= np.arange(3).reshape(1,3)
arr2= np.arange(12).reshape(4,3)
print(arr1)
print(arr2)
```

###### if there's any 1 charo me se... and corresponding bagal wala no is same in both.. (like 3 here), then allowed.

```python
arr1+arr2
```

```python
arr1= np.arange(4).reshape(4,1)
arr2= np.arange(12).reshape(4,3)
print(arr1)
print(arr2)
```

```python
arr1+arr2                          #here also there's one 1 and corresponding bagal wale are same.
```

```python
#if there's a 1 but saamne wale are not same ex- (1,3) and (4,4)/(3,4) then addition not allowed!
```

##### now if there are two 1s and bagal wale same then addition allowed!

```python
arr1= np.arange(4).reshape(4,1)
arr2= np.arange(4).reshape(1,4)
print(arr1)
print(arr2)
```

```python
arr1+arr2                                    
```

##### if there's one 1 element array (singleton) and doosri wali kuch bhi.... then also allowed.

```python
arr1= np.arange(1)
arr2= np.arange(12).reshape(3,4)
print(arr1)
print(arr2)
```

```python
arr1+arr2
```

##### another example where addition by broadcasting is allowed

```python
arr1= np.arange(4)
arr2= np.arange(12).reshape(3,4)
print(arr1)
print(arr2)
```

```python
arr1+arr2
```
