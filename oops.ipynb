Pass by reference

```python
class Person:
  def __init__(self, name, gender):
    self.name = name
    self.gender = gender


def greet(person):
  print(f"Hello, {person.name} You are {person.gender} ")
  p1 = Person('Siddharth', 'M')
  return p1                                   # function returning an object

p = Person('Srishti', 'F')                         # p is a reference to an object of Person class; p itself is not an object
z = greet(p)                                  # passing reference  of object as function argument
print(z)
```

Objects are mutuable

```python
class Person:
  def __init__(self, name, gender):
    self.name = name
    self.gender = gender

def greet(person):
  # changing the instance variable value
  person.name = 'Alice'
  return person


p = Person('Srishti', 'F')
print(id(p))                                   # id of object before
z = greet(p)
print(id(z))                                   # id of object after

#Objects are mutuable! hence proved
```

getter/ setter

```python
class Result:
  def __init__(self, roll, marks):
    self.__clg_id = 101
    self.roll = roll
    self.marks = marks

  def get_id(self):
    print(self.__clg_id)

  def set_id(self):
    self.__clg_id = 100



s1 = Result(1,50)
s1.__clg_id = 100                             # new instance attribute namely __clg_id has been created
                       # but i wanted to make change in actual clg_id, therefore creating functions inside the class which can access clg_id


s1.get_id()
s1.set_id()
s1.get_id()
```

above.. data {clg_id} and methods {get_id} and {set_id} served as encapsulation.
Encapsulation primarily means combining related data and the methods that manipulate that data within a class.

# Static variable vs Instance Variable

```python
# instance variable : object ka variable   (scope different for each)
# static variable : class ka variable     (scope same for all)
```

```python
class Bank:
  def __init__(self, amt, city):
    self.amt = amt
    self.city = city
    self.cid = 0
    self.cid+=1

p1 = Bank(400, 'Indore')
print(p1.cid)

p2 = Bank(500, 'Jaipur')
print(p2.cid)

# cid comes out same for both cuz cid is instance variable so, each time it got initialized to 0
```

```python
# below we are making cid as static variable

class Bank:
  cid = 0                                         # static variable
  def __init__(self, amt, city):
    self.amt = amt
    self.cust_id= Bank.cid
    Bank.cid+=1


  # utility methods:
  @staticmethod
  def get_id():                      # static method- without self as parameter
    return Bank.cid

p1 = Bank(400, 'Indore')
print(p1.cid)


p2 = Bank(500, 'Jaipur')
print(p2.cid)

Bank.get_id()
```

Below example shows graceful usage of Public, Private, Protected methods

```python
class BankAccount:
  def __init__(self):
    self.balance = 0

  def _check_balance(self):
    print(f"Your main account balance is: {self.balance} Rs")

  def __update_balance(self, amount):
    self.balance += amount

  def deposit(self, amount):
    if amount>0:
      self.__update_balance(amount)
      self._check_balance()
    else:
      print('Pls enter a valid amount to deposit')


obj1 = BankAccount()
obj1.deposit(450)

obj2 = BankAccount()
obj2._check_balance()

# obj1.__update_balance()       # accessing pvt method directly gives error!
```

```python
# Static Method


class Math:
  def __init__(self, num):
    self.num = num

  @staticmethod
  def add(a,b):
    print(a+b)

obj1 = Math(1)           # creating object

obj1.add(2,3)           # calling static method by object  Note: if you don't write '@staticmethod' over add() then this will give error

Math.add(3,4)           # calling static method by class name
```

```python
# class method

class Company:
  company = 'apple'

  def changeCompany(self, newCompany):
    self.company = newCompany


obj = Company()
obj.changeCompany('Samsung')
print(Company.company)
print(obj.company)
```

```python
# class method

class Company:
  company = 'apple'

  @classmethod
  def changeCompany(cls, newCompany):             # cls is 'class' in place of self which refers to 'object'
    cls.company = newCompany                    #when i use @classmethod decorator, 1st argument to the below function goes as class, otherwise goes as instance


obj = Company()

obj.changeCompany('Samsung')            # calling by object
print(Company.company)
print(obj.company)

Company.changeCompany('Vivo')             # calling by class
print(Company.company)
print(obj.company)
```

getter and setter

```python
class Num:
  def __init__(self, num):
    self._num = num

  def display(self):
    print(self._num)

  @property                                # allows methods to be accessed as if they were attributes
  def ten_times_value(self):
    return 10* self._num

  @ten_times_value.setter
  def ten_times_value(self, new):
    self._num = new



obj1 = Num(2)
obj1.display()
obj1.ten_times_value
obj1.ten_times_value = 40
obj1.display()

```

*args, *kwargs

```python
'''
def fun(normal_arguments, *args, *kwargs)
* always maintain the sequence: normal, args, kwargs
* args and kwargs are not key-words; instead of that any other word (with *) can be taken
* args- is always taken as a tuple.
* kwargs - goes as dictionary
'''
```

```python
def demo(normal, *args):
  print(normal)
  for i in args:
    print(i)
  print("Hello", args[0])
  print(type(args))


norm = "Hello all!"
list1 = ['Srishti', 'Alice', 'Bob', 'Charlie']
demo(norm, *list1)
```

```python
def demo2(**kwargs):
  for k,v in kwargs.items():
    print(k,v)

dic = {'prod_id': 101,
       'prod_type': 'coffee',
       'prod_name': 'Nestle',
       'prod_price': 50}
demo2(**dic)
```

Task:
* Add _fuel_level as a private attribute to Truck.
* Create getter/setter using @property.
* Prevent setting fuel above 100.


```python
class Truck:
  def __init__(self, fuel_level):
    self.__fuel_level = fuel_level

  @property
  def fuel_level(self):
    return self.__fuel_level

  @fuel_level.setter
  def fuel_level(self, new):
    if new <= 100:
      self.__fuel_level = new
      print('Thanks! visit again')
    else:
      print("Overflow warning!! Tank capacity full!!")


t1 = Truck(80)
t1.fuel_level
t1.fuel_level = 500


```
