* PipelineManager – orchestrates the above classes.

```python
import pandas as pd
```

```python
class DataLoader:
    def __init__(self, file):
        self.file = file
        
    def read(self):
        self.df = pd.read_csv(self.file)
        print(self.df.head())

f = DataLoader('index.csv')
f.read()
```

```python
class DataCleaner(DataLoader):
    def count_null(self):
        # super().read()
        return self.df.isnull().sum()

    def handle_null(self):
        return self.df.fillna('unknown',inplace=True)


    def type_checking(self):
        return self.df.dtypes

    def type_conversion(self):
        pass

        

f = DataCleaner('index.csv')
f.read()
print('Null counts')
print(f.count_null())
f.handle_null()
print('--------------------------------------------------------')
print(f.count_null())
print('--------------------------------------------------------')
print('Checking dtypes')
print(f.type_checking())
```
