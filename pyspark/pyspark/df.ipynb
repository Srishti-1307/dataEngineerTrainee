```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName('creating-df').getOrCreate()
```

```python
data = [('Alice', 1, 'A'), ('Bob', 2, 'A'), ('Charlie', 3, 'C'), ('David', 4, 'B'), ('Elena', 5, 'C')]
```

```python
df = spark.createDataFrame(data)
```

```python
df.show()
```

```python
# providing customized col names
```

```python
df = spark.createDataFrame(data, ['name', 'id', 'grade'])
```

```python
df.show()
```

```python
# providing schema
```

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType
```

```python
schema = StructType([
        StructField('Name', StringType(), True),
        StructField('id', IntegerType(), True),
        StructField('grade', StringType(), True)])


# atlast nullable=True Indicates that the column can contain null values.
```

```python
df = spark.createDataFrame(data, schema)
```

```python
# creating df using rdd
```

```python
rdd = spark.sparkContext.parallelize(['a', 'b', 'c', 'c', 'd'])
```

```python
schema = StructType([
        StructField('Values', StringType(), True)])
```

```python
df = rdd.toDF(schema)
```

```python
print(type(df))
```

```python
df.show()
```

```python
data = [('A', 1), ('B', 2), ('C', 3)]

rdd = spark.sparkContext.parallelize(data)
```

```python
df = rdd.toDF(['key', 'val'])
```
