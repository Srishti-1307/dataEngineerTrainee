```python
import pyspark
```

```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName('first spark application').master("local[*]").getOrCreate()
```

```python
df = spark.read.csv('data.csv')
```

```python
df.show()
```

```python
print(type(df))
```

```python
df = spark.read.format('csv').option('inferSchema', True).option('Header', True).load('data.csv')
```

```python
df.show()
```

```python
df.printSchema()
```

```python
df_json = spark.read.format('json').option('inferSchema', True).option('Header', True).option('multiline', False).load('data.json')
```

```python
df_json.show()
```

```python
df_parq = spark.read.format('parquet').option('inferSchema', True).option('Header', True).load('iris.parquet')
```

```python
df_parq.show()
```
