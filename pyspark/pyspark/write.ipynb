```python
from pyspark.sql import  SparkSession
```

```python
spark = SparkSession.builder.appName('writing-to-a-file').getOrCreate()
```

```python
data = [("Alice", "New York", 34),
        ("Bob", "Chicago", 45),
        ("Cathy", "Los Angeles", 29),
       ("David", "Canada", 11),
       ("Elena", "Mexico", 21),
       ("franklin", "Itlay", 13),
       ("George", "Hong Kong", 19),
       ("Hatel", "Sydney", 31)]
```

```python
columns = ["Name", "City", "Age"]
```

```python
df = spark.createDataFrame(data, columns)
```

```python
df.write.format('csv').option('Header', True).mode('overwrite').save('written1.csv')
```

```python
df.coalesce(2).write.format('csv').option('Header', True).mode('overwrite').save('written2.csv')
```

#### if I do not specify no of partitions... then multiple files will get created.. this is because..  Spark operates
in a distributed manner, and each worker writes its part of the data in parallel. 
The _SUCCESS file is an empty, automatically generated marker file that confirms the Spark job completed successfully.
